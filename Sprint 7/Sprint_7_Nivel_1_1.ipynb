Ejercicio 1: Selección de Personal

```python
# Esta biblioteca permite imprimir diccionarios en vertical para una mejor visualización.
import pprint

# La función de validación de datos comprueba si cada uno de los datos
# introducidos se ajusta al formato de datos esperado.
def validar_dades(candidato):
    condiciones = [
        candidato["Nombre"].isalpha(),
        candidato["Apellido"].isalpha(),
        candidato["Teléfono"].isnumeric() and len(candidato["Teléfono"]) == 9,
        0 <= candidato["Experiencia"] <= 50,
        0 <= candidato["Nivel de Catalan"] <= 3,
        0 <= candidato["Nivel de Excel"] <= 3,
        0 <= candidato["Título académico"] <= 3,
    ]
    return all(condiciones)

# La función calcular_puntuacio() recibe un candidato y devuelve su puntuación:
# años de experiencia * 1.5 más los puntos de catalán, Excel y título académico.
def calcular_puntuacio(candidato):
    return (candidato["Experiencia"] * 1.5
            + candidato["Nivel de Catalan"]
            + candidato["Nivel de Excel"]
            + candidato["Título académico"])

# La función classificar_candidat() asigna una clasificación según el rango
# de puntuación de cada candidato y devuelve una nueva lista que contiene
# únicamente el ID, nombre, apellido, teléfono, puntuación y clasificación.
def classificar_candidat(datos):
    classificacion = []
    for candidato in datos:
        if candidato["Puntuación"] >= 12:
            clas = "Prioridad Alta"
        elif candidato["Puntuación"] >= 8:
            clas = "Candidato adecuado"
        else:
            clas = "No prioritario"

        candidato_clasif = {
            "ID": candidato["ID"],
            "Nombre": candidato["Nombre"],
            "Apellido": candidato["Apellido"],
            "Teléfono": candidato["Teléfono"],
            "Puntuación": candidato["Puntuación"],
            "Clasificación": clas
        }
        classificacion.append(candidato_clasif)
    return classificacion

# La función mostrar_individual() muestra el nombre y la puntuación de un
# candidato en el momento en que se registra.
def mostrar_individual(candidato):
    print(f"\nCandidato válido registrado: {candidato['Nombre']} {candidato['Apellido']}")
    print(f"Puntuación: {candidato['Puntuación']}")

# La función mostrar_informe() ordena los candidatos de mayor a menor
# puntuación, los clasifica con classificar_candidat() e imprime el informe
# final con pprint.pprint().
def mostrar_informe(registro):
    registro_ordenado = sorted(registro, key=lambda c: c["Puntuación"], reverse=True)
    print("\n---------Informe de selección---------\n")
    pprint.pprint(classificar_candidat(registro_ordenado), sort_dicts=False)

# La función obtenir_dades() recorre los datos de ejemplo, valida a cada candidato,
# calcula su puntuación, le asigna un ID y lo guarda en una lista de diccionarios.
# Deja de aceptar candidatos al llegar a 10 registros válidos.
def obtenir_dades():

    registro_candidatos = []

    registro_candidatos_test = [
        "Laura Serra", "608112233", 6, 3, 3, 3,
        "Clara Gómez", "655221199", 3, 3, 3, 2,
        "Sergio López", "644112211", 3, 2, 2, 2,
        "Candidat Error", "611000000", 5, 7, 3, 2,  # Català = 7, NO VÁLIDO
        "Marta Ruiz", "699887766", 2, 3, 2, 1,
        "Marc Vidal", "612443322", 6, 3, 3, 3,
        "Júlia Romero", "677998877", 5, 3, 3, 2,
        "Pol Navarro", "699554433", 4, 3, 2, 2,
        "David Ferrer", "677665544", 2, 1, 1, 1,
        "Anna Paredes", "633221144", 1, 2, 1, 0,
        "Anibal Paredes", "644272727", 1, 2, 1, 0,
        "Núria Casals", "600112200", 8, 2, 3, 3  # 11º candidato
    ]

    tamano_registro = 6
    num_max_candidatos = 10

    for i in range(0, len(registro_candidatos_test), tamano_registro):
        datos_candidato = registro_candidatos_test[i:i + tamano_registro]
        nombre, apellido = datos_candidato[0].split()

        candidato = {
            "ID": len(registro_candidatos) + 1,
            "Nombre": nombre,
            "Apellido": apellido,
            "Teléfono": datos_candidato[1],
            "Experiencia": int(datos_candidato[2]),
            "Nivel de Catalan": int(datos_candidato[3]),
            "Nivel de Excel": int(datos_candidato[4]),
            "Título académico": int(datos_candidato[5]),
            "Puntuación": 0
        }

        if validar_dades(candidato):
            candidato["Puntuación"] = calcular_puntuacio(candidato)
            registro_candidatos.append(candidato)
            mostrar_individual(candidato)
        else:
            print(f"\nCandidato inválido, no registrado: {nombre} {apellido}")

        # Al llegar a 10 candidatos válidos se dejan de aceptar registros.
        if len(registro_candidatos) == num_max_candidatos:
            print(f"\nSe han inscrito {num_max_candidatos} candidatos. No se admiten más candidaturas.")
            break

    return registro_candidatos

# Finalmente, se define la función main_seleccio(), que llama a las funciones
# obtenir_dades() y mostrar_informe() para generar el informe de selección final.
def main_seleccio():
    registro = obtenir_dades()
    mostrar_informe(registro)

# Aquí se ejecuta la función principal, main_seleccio().
main_seleccio()
```

Alogoritmo:

3. ¿Qué clientes presentan tan poco interés que no vale la pena invertir presupuesto?

```python
# Esta biblioteca permite crear tablas y agrupar datos.
import pandas as pd

# Base de datos proporcionada.
clients = [
    {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
    {"id_client": 2, "nom": "Marc Vidal", "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
    {"id_client": 3, "nom": "Ana López", "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
    {"id_client": 4, "nom": "Joan Riera", "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
    {"id_client": 5, "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
    {"id_client": 6, "nom": "Luis García", "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
    {"id_client": 7, "nom": "Pilar Sánchez", "edat": 61, "grup_edat": "60+", "codi_comportament": 3},
    {"id_client": 8, "nom": "Eva Martín", "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
    {"id_client": 9, "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
    {"id_client": 10, "nom": "Núria Costa", "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
]

# Riesgo final por debajo del cual no vale la pena invertir presupuesto.
UMBRAL_BAJO_INTERES = 0.30
# Código de comportamiento de los clientes que llegan al checkout pero no pagan.
CODIGO_CHECKOUT = 2


# La siguiente función recibe como argumento la lista de clientes y devuelve una nueva
# lista de diccionarios con los factores de comportamiento, de edad y el riesgo final
# calculado de cada cliente.
def processar_clients(clients):
    result_segments = []
    for cliente in clients:
        segment_info = obtenir_segment_i_factor(cliente["codi_comportament"])
        edad_info = obtenir_factor_edat(cliente["grup_edat"])
        riesgo = calcular_risc(segment_info["factor"], edad_info["factor"])

        datos_cliente = {
            "id_cliente": cliente["id_client"],
            "nombre": cliente["nom"],
            "edad": cliente["edat"],
            "grupo_edad": cliente["grup_edat"],
            "codigo_comportamiento": cliente["codi_comportament"],
            "segmento": segment_info["segmento"],
            "mensaje": segment_info["mensaje"],
            "factor_codigo": segment_info["factor"],
            "factor_edad": edad_info["factor"],
            "riesgo_final": riesgo
        }

        result_segments.append(datos_cliente)
    return result_segments


# Esta función recibe como argumento el código de comportamiento de cada cliente
# y funciona como un diccionario de búsqueda donde cada índice corresponde a un código,
# devolviendo el segmento, el mensaje y el factor asociados.
def obtenir_segment_i_factor(codi_comportament):

    segments = {
        0: {"segmento": "Navega pero no añade productos",
            "factor": 0.1, "mensaje": "Tu carrito de compras está vacío, ¡vamos a agregar algunos artículos!"},
        1: {"segmento": "Agregar productos pero no llega al checkout",
            "factor": 0.25, "mensaje": "Todavía hay artículos en tu carrito pendientes de checkout, ¡no te olvides de ellos!"},
        2: {"segmento": "Llega al checkout pero no paga",
            "factor": 0.5, "mensaje": "¡Ya casi lo tienes, solo falta pagar!"},
        3: {"segmento": "Cesta de compra abandonado hace menos de 24 h",
            "factor": 0.75, "mensaje": "¡Tu cesta de la compra te está esperando!"},
        4: {"segmento": "Cesta de compra abandonado hace más de 24 h",
            "factor": 0.9, "mensaje": "¡Tu cesta de la compra lleva esperándote más de 24 horas, no te rindas!"}
    }

    return segments[codi_comportament]


# Esta función recibe como argumento el grupo de edad de cada cliente y funciona
# como un diccionario de búsqueda donde cada índice corresponde a un rango de edad,
# devolviendo el factor asociado.
def obtenir_factor_edat(grup_edat):

    factor_edad = {
        "18-25": {"factor": 1.2},
        "26-40": {"factor": 1.0},
        "41-60": {"factor": 0.85},
        "60+": {"factor": 0.7}
    }
    return factor_edad[grup_edat]


# Esta función recibe como argumento el factor de código y el factor de edad de
# un cliente, los multiplica y devuelve el valor del riesgo final.
def calcular_risc(factor_codigo, factor_edad):
    return round(factor_codigo * factor_edad, 2)


# La siguiente función convierte en un dataframe la lista de diccionarios devuelta por
# processar_clients(), ordena de forma descendente los valores de riesgo_final y
# devuelve el dataframe generado.
def mostrar_dataframe(df_input):
    df = pd.DataFrame(df_input)
    df_ordenado = df.sort_values(["riesgo_final"], ascending=[False]).reset_index(drop=True)
    return df_ordenado


# Esta función responde a las preguntas de la campaña de retargeting a partir de
# los datos del dataframe. Las conclusiones se calculan con los datos, no están escritas a mano.
def mostrar_respostes(df):
    columnas = ["id_cliente", "nombre", "riesgo_final"]

    # 1. ¿Qué clientes deben contactar primero en una campaña de retargeting?
    print("------------------------------Campaña de retargeting------------------------------")
    print("Necesitamos contactar primero a los 3 clientes con mayor riesgo final: \n")
    print(df[columnas].head(3))

    # 2. ¿Qué diferencias existen entre clientes con el mismo comportamiento
    # pero de edades distintas?
    print("----------------------------------------------------------------------------------")
    print("Riesgo final de los clientes con el mismo comportamiento, de menor a mayor edad: \n")
    comparacion = df.sort_values(["codigo_comportamiento", "edad"])
    print(comparacion[["codigo_comportamiento", "nombre", "edad", "factor_edad", "riesgo_final"]].to_string(index=False))

    # Ejemplo con el código 4: cesta abandonada hace más de 24 h.
    grupo = comparacion[comparacion["codigo_comportamiento"] == 4]
    joven, mayor = grupo.iloc[0], grupo.iloc[-1]
    print(f"\nEjemplo (código 4): {joven['nombre']} ({joven['edad']} años) tiene un riesgo de "
          f"{joven['riesgo_final']} y {mayor['nombre']} ({mayor['edad']} años) tiene {mayor['riesgo_final']}.")
    print("Con el mismo comportamiento el riesgo no es igual: el factor de edad es mayor "
          "cuanto más joven es el cliente.")

    # 3. ¿Qué clientes presentan tan poco interés que no vale la pena invertir presupuesto?
    print("----------------------------------------------------------------------------------")
    print(f"No merece la pena incluir en la campaña a los clientes con riesgo final "
          f"de {UMBRAL_BAJO_INTERES} o menos: \n")
    print(df.loc[df["riesgo_final"] <= UMBRAL_BAJO_INTERES, columnas])

    # 4. ¿Qué cliente sería el más rentable de recuperar?
    print("----------------------------------------------------------------------------------")
    checkout = df[df["codigo_comportamiento"] == CODIGO_CHECKOUT]
    mejor = checkout.loc[checkout["riesgo_final"].idxmax()]
    print("Los clientes que llegan al checkout pero no pagan son los más cercanos a comprar. \n"
          "Entre ellos, el más rentable de recuperar es el de mayor riesgo final: \n")
    print(f"{mejor['nombre']} (id {mejor['id_cliente']}), riesgo final {mejor['riesgo_final']}")


# Por último, esta función recibe como argumento la base de datos proporcionada y orquesta
# la ejecución de las demás funciones.
def main_marketing(clients):

    df_input = processar_clients(clients)
    df_ordenado = mostrar_dataframe(df_input)
    mostrar_respostes(df_ordenado)
    print("\nDatos de los clientes: \n")
    return df_ordenado


main_marketing(clients)
```

Ejercicio 1: Gestión de riesgo y priorización de pacientes

```python
# Esta biblioteca permite imprimir diccionarios en vertical para una mejor visualización.
import pprint
# Estas bibliotecas permiten usar funciones matemáticas, de agrupación y creación de tablas.
import pandas as pd
import numpy as np
# Base de datos proporcionada.
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho", #  3 + 4 + 1 = 8
    "Luis|33|dolor_cabeza,faTiga", # 1 + 1 = 2
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos", # 3 + 5 + 1 = 9
    "Jordi|52|fatiga , tos", # 1 + 1 = 2
    "Anna|29|fiebre alta,dolor_cabeza" # 3 + 1 = 4
]
# Esta función asocia el valor de riesgo a los síntomas de cada paciente y calcula
# el valor total del riesgo.
def riesgos(sintomas):
  nivells_risc = {
      "fiebre_alta": 3,
      "dificultad_respirar": 5,
      "dolor_pecho": 4,
      "tos": 1,
      "fatiga": 1,
      "dolor_cabeza": 1
      }
  valor_riesgo = 0
  for riesgo in nivells_risc:
    if riesgo in sintomas:
      valor_riesgo = valor_riesgo + nivells_risc[riesgo]
  return valor_riesgo

# Esta función convierte la lista de pacientes en un dataframe ordenado por el valor del
# riesgo total de cada paciente.
def mostrar_dataframe(lista_pacients):
  df = pd.DataFrame(lista_pacients)
  df_ordenado = df.sort_values(["riesgo_total"], ascending=[False])
  return df_ordenado

# Esta función convierte la lista de strings con la información de cada paciente en
# una lista de diccionarios y normaliza los datos para poder compararlos
# con los valores predeterminados de los síntomas listados en la base de datos.
def convertir_y_normalizar_datos(pacients):
  lista_pacients = []
  for i in range(len(pacients)):
    pacients[i] = pacients[i].replace("|",",")
    pacients[i] = pacients[i].split(",")

    if len(pacients[i]) == 4:
        convertido = {
          "Id": 0,
          "nombre": (pacients[i][0]).strip(),
          "edad": (pacients[i][1]).strip(),
          "sintomas normalizados": [s.strip().lower().replace(" ", "_") for s in pacients[i][2:]],
          "riesgo_total": 0
          }
        lista_pacients.append(convertido)
    else:
        convertido = {
          "Id": 0,
          "nombre": (pacients[i][0]).strip(),
          "edad": (pacients[i][1]).strip(),
          "sintomas normalizados": [s.strip().lower().replace(" ", "_") for s in pacients[i][2:]],
          "riesgo_total": 0
          }
        lista_pacients.append(convertido)
  return lista_pacients

# La función principal orquesta las demás funciones, establece un id para cada paciente
# y devuelve el dataframe procesado al final.
def main_pacientes(pacients):
  lista_pacients = convertir_y_normalizar_datos(pacients)
  contador_id = 0
  for paciente in lista_pacients:
    contador_id += 1
    paciente["Id"] = contador_id
    paciente["riesgo_total"] = riesgos(paciente["sintomas normalizados"])
  df_pacientes = mostrar_dataframe(lista_pacients)
  print("Lista de pacientes ordenados por el riesgo total:\n")
  return df_pacientes

main_pacientes(pacients)
```

Ejercicio 2: Clasificación de clientes de seguros

```python
# Esta biblioteca permite imprimir diccionarios en vertical para una mejor visualización.
import pprint
# Esta biblioteca permite usar funciones estadísticas.
import scipy.stats as stats
# Estas bibliotecas permiten usar funciones matemáticas, de agrupación y creación de tablas.
import pandas as pd
import numpy as np
# Base de datos proporcionada.

clients = [
    {"nombre": "Ana", "edad": 42, "codigo": 0},  # precio final = 120 * (1 + 42/100) = 170.4
    {"nombre": "Carlos", "edad": 30, "codigo": 2}, # precio final = 350 * (1 + 30/100) = 455
    {"nombre": "Isabel", "edad": 55, "codigo": 3}, # precio final = 500 * (1 + 55/100) = 775
    {"nombre": "Jorge", "edad": 40, "codigo": 1}, # precio final = 200 * (1 + 40/100) = 280
    {"nombre": "Marta", "edad": 28, "codigo": 0} # precio final = 120 * (1 + 28/100) = 153.6
]

# Esta función asocia el valor del código del seguro de cada cliente con la categoría,
# el precio base y la probabilidad de fraude, y lo devuelve según el cliente elegido.
def obtener_nivel(codigo):

  nivells_seguro = {
    0: {"categoria": "Riesgo bajo", "precio_base": 120, "prob_fraude": 0.01},
    1: {"categoria": "Riesgo medio", "precio_base": 200, "prob_fraude": 0.05},
    2: {"categoria": "Riesgo alto", "precio_base": 350, "prob_fraude": 0.15},
    3: {"categoria": "Riesgo crítico", "precio_base": 500, "prob_fraude": 0.30}
    }
  return nivells_seguro[codigo]

# Esta función calcula el precio final de cada cliente y añade a cada registro
# el elemento precio_final, devolviendo la lista actualizada de clientes.
def precio_seguro(clients):
  for cliente in clients:
    precio_base = cliente["precio_base"]
    edad = cliente["edad"]
    precio_final = precio_base * (1 + edad/100)
    cliente.update({"precio_final": precio_final})
  return(clients)

# Esta función convierte la lista de registros en un dataframe, organiza la disposición de las columnas
# y ordena las filas a partir del valor de precio_base.
def mostrar_dataframe(clients):

  df = pd.DataFrame(clients)
  df_ordenado = df.sort_values(["precio_base"], ascending=[False])
  df_ordenado = df_ordenado[["Id", "nombre", "edad", "codigo", "categoria", "prob_fraude", "precio_base", "precio_final"]]
  return df_ordenado

# Esta función realiza un análisis estadístico simplificado de las relaciones entre edad,
# precio y probabilidad de fraude para cada cliente.
def analisis_seguro(df):
  corr_edad, _ = stats.pearsonr (df["edad"], df["prob_fraude"])
  corr_precio_final, _ = stats.pearsonr (df["precio_final"], df["prob_fraude"])
  print("Correlación probabilidad de fraude x edad (R):" , round(corr_edad, 2))
  print("Correlación probabilidad de fraude x precio final (R):" , round(corr_precio_final, 2))
  print("El siguiente cliente presentó la mayor probabilidad de cometer fraude, \ny se observa a través de la correlación de Pearson (R) que este es precisamente \nel que tiene el precio final del seguro más alto y, por lo tanto, debe ser revisado manualmente.\n\n", df.iloc[0])


# La función principal orquesta las demás funciones y devuelve el análisis del cliente
# más crítico y el dataframe de los clientes ordenados por criticidad.
def main_seguros(clients):

  contador_id = 0
  for cliente in clients:
    contador_id += 1
    cliente["Id"] = contador_id
    nivel = obtener_nivel(cliente["codigo"])
    cliente.update(nivel)
  precio_seguro(clients)
  mostrar_dataframe(clients)
  df = mostrar_dataframe(clients)
  analisis_seguro(df)
  print("\n\nLista de clientes oredenados por riesgo de fraude\n")
  return df

main_seguros(clients)
```

Ejercicio 1: Generador de contraseñas seguras

```python
# Esta biblioteca permite usar funciones matemáticas y, en este código en especial,
# el módulo random.
import numpy as np
# Esta biblioteca permite usar funciones para la manipulación de strings.
import string


# Esta función hace una pregunta de S/N y la repite hasta recibir una respuesta válida.
# Devuelve True si la respuesta es "S" y False si es "N".
def preguntar_sn(pregunta):
    while True:
        respuesta = input(pregunta + " S/N\n").strip().upper()
        if respuesta == "S":
            return True
        if respuesta == "N":
            return False
        print("¡Entrada inválida!\n")


# Esta función pregunta al usuario qué tipos de caracteres quiere en la contraseña.
# Devuelve un diccionario con los parámetros elegidos, o un diccionario vacío si el
# usuario no quiere modificarlos (así se mantienen los valores por defecto de
# generar_contrasena()). Si no se elige ningún tipo, vuelve a preguntar.
def pedir_parametros():
    if not preguntar_sn("¿Desea modificar los parámetros de la contraseña?"):
        print("Parámetros mantenidos.\n")
        return {}

    while True:
        parametros = {
            "majuscules": preguntar_sn("¿Desea incluir letras mayúsculas en la contraseña?"),
            "minuscules": preguntar_sn("¿Desea incluir letras minúsculas en la contraseña?"),
            "numeros": preguntar_sn("¿Desea incluir números en la contraseña?"),
            "signes": preguntar_sn("¿Desea incluir símbolos en la contraseña?"),
        }
        if any(parametros.values()):
            return parametros
        print("Al menos un parámetro tiene que ser elegido.\n")


# Esta función recibe los cuatro parámetros y devuelve la lista de grupos de caracteres
# que se pueden usar (mayúsculas, minúsculas, números y/o símbolos).
def obtener_grupos(majuscules, minuscules, numeros, signes):
    grupos = []
    if majuscules:
        grupos.append(string.ascii_uppercase)
    if minuscules:
        grupos.append(string.ascii_lowercase)
    if numeros:
        grupos.append(string.digits)
    if signes:
        grupos.append(string.punctuation)
    return grupos


# Función principal del generador. Los valores por defecto de los cuatro primeros
# parámetros son los que pide el enunciado. Si longitud es None, se elige al azar
# un tamaño entre 6 y 9 caracteres.
def generar_contrasena(majuscules=True, minuscules=True, numeros=True, signes=False, longitud=None):
    grupos = obtener_grupos(majuscules, minuscules, numeros, signes)
    if not grupos:
        raise ValueError("Al menos un tipo de carácter tiene que estar activado.")

    if longitud is None:
        longitud = np.random.randint(6, 10)
    if longitud < len(grupos):
        raise ValueError(f"La longitud mínima para esta configuración es {len(grupos)}.")

    # Todos los caracteres posibles según los parámetros elegidos.
    alfabeto = np.array(list("".join(grupos)))
    # Un carácter elegido al azar de cada grupo activo, para asegurar que todos aparecen.
    obligatorios = np.array([np.random.choice(list(g)) for g in grupos])
    # El resto de la contraseña se completa con caracteres del alfabeto completo.
    restantes = np.random.choice(alfabeto, size=longitud - len(grupos))

    # Se juntan los caracteres obligatorios con los restantes, se mezclan al azar
    # y se convierten en un string final.
    return "".join(np.random.permutation(np.concatenate([obligatorios, restantes])))


# La función principal orquesta las demás funciones y devuelve la contraseña generada.
def main_contrasena():
    parametros = pedir_parametros()
    return generar_contrasena(**parametros)


print("Contraseña generada:", main_contrasena())
```

Ejercicio 2: Procesamiento automático de datos deportivos

```python
# Esta biblioteca permite realizar peticiones HTTP para obtener datos desde una URL.
import requests

# Esta biblioteca permite trabajar con DataFrames y realizar análisis de datos.
import pandas as pd


# Esta función descarga los datos históricos de los partidos desde una URL.
# Devuelve una lista donde cada elemento corresponde a una línea del archivo.
def importar_datos():

  url = "https://raw.githubusercontent.com/aregueiraaraujo/Data-Analytics/main/Sprint%207/historic%20partits.txt"
  resposta = requests.get(url)
  resposta.raise_for_status()
  lineas = resposta.text.splitlines()
  return lineas


# Esta función recibe la lista de partidos y separa los datos de cada partido.
# Cada partido contiene el equipo local, el resultado y el equipo visitante.
# Los goles se convierten de texto a números enteros.
# Devuelve una lista de diccionarios con los datos de cada partido normalizados.
def filtrar_datos(partidas):

  partidas_normalizado = []
  for partido in partidas:
    local, resultado, visitante = partido.split('\t')
    goles_local, goles_visitante = map(int, resultado.split('-'))

    partidas_normalizado.append({
        'local': local,
        'goles_local': goles_local,
        'goles_visitante': goles_visitante,
        'visitante': visitante
    })

  return partidas_normalizado


# Esta función recibe los partidos normalizados y calcula las estadísticas
# acumuladas de cada equipo.
# Se registran los partidos jugados, ganados, empatados y perdidos,
# los goles a favor, los goles en contra y los puntos obtenidos.
def calcular_estadisticas(partidas):

  estadisticas = {}

  for partido in partidas:

    # Se obtienen los datos de cada partido.
    local = partido['local']
    visitante = partido['visitante']
    goles_local = partido['goles_local']
    goles_visitante = partido['goles_visitante']


    # Si el equipo local todavía no existe en el diccionario,
    # se crea su registro con todas las estadísticas inicializadas en cero.
    if local not in estadisticas:
      estadisticas[local] = {
          'Partidos Jugados': 0,
          'Partidos Ganados': 0,
          'Empates': 0,
          'Partidos Perdidos': 0,
          'Goles a Favor': 0,
          'Goles en Contra': 0,
          'Puntos': 0
      }

    # Si el equipo visitante todavía no existe en el diccionario,
    # se crea su registro con todas las estadísticas inicializadas en cero.
    if visitante not in estadisticas:
      estadisticas[visitante] = {
          'Partidos Jugados': 0,
          'Partidos Ganados': 0,
          'Empates': 0,
          'Partidos Perdidos': 0,
          'Goles a Favor': 0,
          'Goles en Contra': 0,
          'Puntos': 0
      }

    # Cada equipo suma un partido jugado.
    estadisticas[local]['Partidos Jugados'] += 1
    estadisticas[visitante]['Partidos Jugados'] += 1

    # Se suman los goles marcados por cada equipo.
    estadisticas[local]['Goles a Favor'] += goles_local
    estadisticas[visitante]['Goles a Favor'] += goles_visitante

    # Se suman los goles recibidos por cada equipo.
    estadisticas[local]['Goles en Contra'] += goles_visitante
    estadisticas[visitante]['Goles en Contra'] += goles_local

    # Se compara el número de goles para determinar el resultado del partido.
    # Resultado
    if goles_local > goles_visitante:

      # El equipo local gana y recibe 3 puntos.
      estadisticas[local]['Partidos Ganados'] += 1
      estadisticas[local]['Puntos'] += 3

      # El equipo visitante pierde.
      estadisticas[visitante]['Partidos Perdidos'] += 1

    elif goles_local < goles_visitante:

      # El equipo visitante gana y recibe 3 puntos.
      estadisticas[visitante]['Partidos Ganados'] += 1
      estadisticas[visitante]['Puntos'] += 3

      # El equipo local pierde.
      estadisticas[local]['Partidos Perdidos'] += 1

    else:

      # Si ambos equipos marcan el mismo número de goles,
      # el partido termina en empate.
      estadisticas[local]['Empates'] += 1
      estadisticas[visitante]['Empates'] += 1

      # Cada equipo recibe 1 punto por el empate.
      estadisticas[local]['Puntos'] += 1
      estadisticas[visitante]['Puntos'] += 1

  return estadisticas


# Esta función recibe las estadísticas de todos los equipos y las convierte
# en un DataFrame de pandas para facilitar su visualización y análisis.
def crear_dataframe(estadisticas):

  dataframe = pd.DataFrame.from_dict(
      estadisticas,
      orient='index'
  )

  # Se transforma el índice del DataFrame en una columna.
  dataframe = dataframe.reset_index()

  # Se cambia el nombre de la columna que contiene los nombres de los equipos.
  dataframe = dataframe.rename(columns={'index': 'equipo'})

  # Se calcula restando los goles en contra a los goles a favor.
  dataframe['Diferencia de Goles'] = dataframe['Goles a Favor'] - dataframe['Goles en Contra']

  # Los equipos se ordenan primero por puntos, después por diferencia de goles
  # y finalmente por goles a favor.
  dataframe = dataframe.sort_values(
      by=['Puntos', 'Diferencia de Goles', 'Goles a Favor'],
      ascending=[False, False, False]
  )

  # Crear posición
  # Se añade una columna con la posición de cada equipo en la clasificación.
  dataframe.insert(
      0,
      'posicion',
      range(1, len(dataframe) + 1)
  )

  return dataframe


# Esta función principal coordina todas las funciones anteriores.
# Primero importa los datos, después los filtra, calcula las estadísticas
# y finalmente crea el DataFrame con la clasificación.
def main_times():

  datos_importados = importar_datos()
  partidas = filtrar_datos(datos_importados)
  estadisticas = calcular_estadisticas(partidas)
  dataframe = crear_dataframe(estadisticas)

  return dataframe


# Se ejecuta la función principal.
main_times()
```
