```python
import requests
import json
from pprint import pprint
```

```python
posts = requests.get("https://jsonplaceholder.typicode.com/posts")
status_post = posts.status_code
if status_post == 200:   
    posts = posts.json()
    tamano_posts = len(posts)
    print("Tamaño de la lista de publicaciones: ",tamano_posts)
    print("Status code: ",status_post)
else: 
    print("Error al obtener las publicaciones") 
    
users = requests.get("https://jsonplaceholder.typicode.com/users")
status_user = users.status_code
if status_user == 200:
    users = users.json()
    tamano_users = len(users)
    print("Tamaño de la lista de usuarios: ",tamano_users)
    print("Status code: ",status_user)
else: 
    print("Error al obtener los usuarios") 
    
todos = requests.get("https://jsonplaceholder.typicode.com/todos")
status_task = todos.status_code
if status_task == 200:
    todos = todos.json()
    tamano_todos = len(todos)
    print("Tamaño de la lista de tareas: ",tamano_todos)
    print("Status code: ",status_task)
else: 
    print("Error al obtener las tareas") 
```

```python
url_post = "https://jsonplaceholder.typicode.com/posts"
nuevo_post = {
    "title": "Título de la Publicación Fictícia",
    "body": "Este es el cuerpo de mi nueva publicación de prueba.",
    "userId": 1
}
response_post = requests.post(url_post, json=nuevo_post)
print("--- POST ---")
print(f"Código de estado: {response_post.status_code}") 
print("Respuesta JSON:")
print(json.dumps(response_post.json(), indent=2))
print("\n")
```

```python
url_patch = "https://jsonplaceholder.typicode.com/posts/1"
modificacion = {
    "title": "Título modificado parcialmente mediante PATCH"
}
response_patch = requests.patch(url_patch, json=modificacion)
print("--- PATCH ---")
print(f"Código de estado: {response_patch.status_code}")
print("Resposta JSON:")
print(json.dumps(response_patch.json(), indent=2))
print("\n")
```

```python
url_delete = "https://jsonplaceholder.typicode.com/posts/1"
response_delete = requests.delete(url_delete)
print("--- DELETE ---")
print(f"Código de estado: {response_delete.status_code}")
print("Respuesta JSON:")
print(json.dumps(response_delete.json(), indent=2))
```

/v1/forecast: Endpoint principal para obtener datos meteorológicos actuales, horarios y diarios.
/v1/historical: Endpoint para consultar el archivo de datos meteorológicos históricos.

timezone: Permite ajustar las fechas y horas a la zona horaria local.
past_days: Añade al pronóstico los datos reales de los días anteriores.
temperature_unit: Permite alternar la unidad de medida de la temperatura.
hourly y daily: Son filtros flexibles donde especificas exactamente qué variables quieres descargar, separadas por comas.

```python
import requests
import pandas as pd
```

```python
url_weather = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 41.40251223414462,
    "longitude": 2.1943790113499237,
    "hourly": ["temperature_2m", "wind_speed_10m"],
    "format": "json" 
}
response = requests.get(url_weather, params=params)
codigo = response.status_code
print("Status code: ",codigo)
if response.status_code == 200:
    datos = response.json()
    lat = datos['latitude']
    lon = datos['longitude']
    elev = datos.get('elevation', 'N/A')
    tz_offset = datos.get('utc_offset_seconds', 0)
    print(f"Coordinates: {abs(lat):.2f}°{'S' if lat < 0 else 'N'} {abs(lon):.2f}°{'W' if lon < 0 else 'E'}")
    print(f"Elevation: {elev} m asl")
    print(f"Timezone difference to GMT+0: {tz_offset}s")
    hourly_data = datos['hourly']
    hourly_dataframe = pd.DataFrame(hourly_data)
    hourly_dataframe['time'] = pd.to_datetime(hourly_dataframe['time'])
    print("\nHourly data\n", hourly_dataframe.head(10))
else:
    print(f"Error. Código del status: {response.status_code}")
```

```python
url_search = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"

parametros = {
    "q": "playa",
    "rows": 3
}

respuesta = requests.get(url_search, params=parametros)
codigo = respuesta.status_code

print(f"Status code: {codigo}\n")

if codigo == 200:
    datos_json = respuesta.json()
    total_encontrados = datos_json["result"]["count"]
    print(f"Se han encontrado {total_encontrados} datasets relacionados con '{parametros['q']}'.\n")
    paquetes = datos_json["result"]["results"]

    for paquete in paquetes:
        print(f"Título: {paquete.get('title')}")
        print(f"Nombre/ID: {paquete.get('name')}")
        print(f"Modificado: {paquete.get('metadata_modified')}")
        recursos = paquete.get("resources", [])
        print(f"Archivos disponibles ({len(recursos)}):")
        
        for recurso in recursos:
            nombre_archivo = recurso.get("name")
            formato = recurso.get("format")
            res_id = recurso.get("id")
            print(f"     - [{formato}] {nombre_archivo} | ID: {res_id}")
            
        print("-" * 50)
else:
    print("Error al realizar la búsqueda.")
```

```python
resource_id = "cc4a6b67-4169-45ae-9f5a-e464908d2417"
url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"
params = {
    "resource_id": resource_id,
    "limit": 100
}
respuesta = requests.get(url_datastore, params=params)

if respuesta.status_code == 200:
    datos_json = respuesta.json()
    registros = datos_json["result"]["records"]
    df = pd.DataFrame(registros)
    print(f"Se han recuperado {len(df)} registros exitosamente.")
    print("\nVista previa de los datos:")
    print(df.head(10))
    nombre_archivo = "datos_opendata_bcn.csv"
    df.to_csv(nombre_archivo, index=False, encoding='utf-8-sig')
    
    print(f"\nArchivo guardado correctamente como: '{nombre_archivo}'")
    
else:
    print(f"Error al realizar la petición. Código de estado: {respuesta.status_code}")
```
